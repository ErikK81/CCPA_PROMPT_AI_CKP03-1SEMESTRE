# CKP03 — Chatbot de Dúvidas de Python

**Tema:** assistente que explica conceitos básicos de Python para iniciantes.

**Integrantes:** preencher nomes e RMs.

## Persona e system prompt
O chatbot se chama **PyAjuda**. É um professor chatbot, fala em português simples, dá exemplos curtos e responde apenas a dúvidas sobre Python. Quando o assunto não for Python, explica educadamente que esse não é seu tema. Não inventa resultados de código. Essas regras tornam as respostas consistentes e fáceis de entender.

**Modelo:** `gpt-oss:120b`.


```python
# Importa as ferramentas para acessar a API e ler o segredo do Colab.
import requests
from google.colab import userdata

CHAVE = userdata.get("OLLAMA_API_KEY")
URL = "https://ollama.com/api/chat"
MODELO = "gpt-oss:120b"

# Define as regras de comportamento do chatbot.
PERSONA = ("Você é PyAjuda, professor de Python para iniciantes. "
           "Responda em português simples, com exemplos curtos. "
           "Fale apenas sobre Python; recuse educadamente outros assuntos. "
           "Se não souber, admita. Não invente execuções de código.")

# Inicia uma conversa nova com a mensagem de sistema.
def nova_conversa():
    return [{"role": "system", "content": PERSONA}]

# Envia todo o histórico para o modelo, preservando o contexto.
def perguntar(historico, pergunta):
    historico.append({"role": "user", "content": pergunta})
    resposta = requests.post(URL, headers={"Authorization": f"Bearer {CHAVE}"},
        json={"model": MODELO, "messages": historico, "stream": False},
        timeout=180)
    resposta.raise_for_status()
    texto = resposta.json()["message"]["content"]
    historico.append({"role": "assistant", "content": texto})
    return texto

# Executa três perguntas seguidas e imprime cada resposta.
def demonstrar(nome, perguntas):
    print("\n===", nome, "===")
    historico = nova_conversa()
    for pergunta in perguntas:
        print("\nAluno:", pergunta)
        print("PyAjuda:", perguntar(historico, pergunta))
```

## Três diálogos documentados
Cada exemplo tem **três turnos do aluno**, executados de verdade quando você usar **Executar tudo**.

1. **Caso normal:** conceitos básicos.
2. **Caso difícil:** erro em código e continuidade do contexto.
3. **Fora do escopo:** pergunta não relacionada a Python e retorno ao tema.


```python
# Demonstra o caso normal, com três perguntas relacionadas.
demonstrar("Caso normal", [
    "O que é uma lista em Python?",
    "Como adiciono um elemento nela?",
    "E como mostro o primeiro elemento?"])

# Demonstra a explicação de um erro e a memória do diálogo.
demonstrar("Caso difícil", [
    "Por que print(10 / 0) dá erro?",
    "Como posso evitar esse erro com try e except?",
    "Mostre um exemplo usando o mesmo cálculo."])

# Demonstra como o chatbot trata assuntos fora do seu domínio.
demonstrar("Fora do escopo", [
    "Qual time ganhou o último campeonato?",
    "Então você só explica Python?",
    "Nesse caso, como funciona o comando for?"])
```
